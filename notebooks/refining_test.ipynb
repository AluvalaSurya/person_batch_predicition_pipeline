```python
daily_first["timestamp"] = pd.to_datetime(daily_first["timestamp"])
```

```python
daily_first["first_entry_minutes"] = (daily_first["timestamp"].dt.hour*60 + daily_first["timestamp"].dt.minute)
```

```python
#step 2: extract weekday feature
daily_first["day_of_week"] = daily_first["timestamp"].dt.dayofweek
```

```python
#step 3: sort data 
daily_first = daily_first.sort_values(["dname","sname","timestamp"])
```

```python
#step 4: previous day arrival
daily_first["previous_day_arrival"] = (
    daily_first.groupby(["dname","sname"])["first_entry_minutes"].shift(1)
)
```

```python
daily_first.head()
```

```python
daily_first.tail()
```

```python
cutoff = daily_first["first_entry_minutes"].quantile(0.95) #9:47
daily_first = daily_first[daily_first["first_entry_minutes"] <= cutoff].reset_index(drop=True)
```

```python
daily_first.tail()
```

```python
daily_first["rolling_mean"] = (
    daily_first.groupby(["dname","sname"])["first_entry_minutes"]
    .transform(lambda x:x.rolling(5).mean())
)
```

```python
daily_first["rolling_std"] = (
    daily_first.groupby(["dname","sname"])["first_entry_minutes"]
    .transform(lambda x:x.rolling(5).std())
)
```

```python
col = dc["first_entry_minutes"].dropna()

Q1 = col.quantile(0.25)
Q3 = col.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = dc[
    (dc["first_entry_minutes"] < lower_bound) |
    (dc["first_entry_minutes"] > upper_bound)
]

print("Number of outliers:", len(outliers))
outliers.head()
```

```python
#sequential split
# split_idx = int(len(df)*0.8)
# train_df = df.iloc[:split_idx].reset_index(drop=True)
# test_df = df.iloc[split_idx:].reset_index(drop=True)

train_parts = []
test_parts = []

for student, g in dc.groupby("student_id"):
    split_idx = max(1, int(len(g) * 0.8))  # safety for small groups

    train_parts.append(g.iloc[:split_idx])
    test_parts.append(g.iloc[split_idx:])

train_df = pd.concat(train_parts).reset_index(drop=True)
test_df  = pd.concat(test_parts).reset_index(drop=True)
```

```python
train_df.head()
```

```python
test_df.head()
```

```python
print(len(train_df), len(test_df))
```

```python
X_train = train_df.drop(columns = ["first_entry_minutes"])
y_train = train_df["first_entry_minutes"]

X_test = test_df.drop(columns = ["first_entry_minutes"])
y_test = test_df["first_entry_minutes"]
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train,y_train)
rf_pred = rf_model.predict(X_test) 
```

```python
rftrain_pred = rf_model.predict(X_train)
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)
```

```python
xgtrain_pred = xgb_model.predict(X_train)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

#  MAE 
rf_train = mean_absolute_error(y_train, rftrain_pred)
rf_mae   = mean_absolute_error(y_test, rf_pred)

xg_train = mean_absolute_error(y_train, xgtrain_pred)
xgb_mae  = mean_absolute_error(y_test, xgb_pred)

#  R2 
rf_r2_train = r2_score(y_train, rftrain_pred)
rf_r2_test  = r2_score(y_test, rf_pred)

xgb_r2_train = r2_score(y_train, xgtrain_pred)
xgb_r2_test  = r2_score(y_test, xgb_pred)

print("RF train MAE:", rf_train)
print("RF test MAE :", rf_mae)
print("RF train R2 :", rf_r2_train)
print("RF test R2  :", rf_r2_test)

print("XGB train MAE:", xg_train)
print("XGB test MAE :", xgb_mae)
print("XGB train R2 :", xgb_r2_train)
print("XGB test R2  :", xgb_r2_test)
```

```python
#hyperparameter tuning
rf_params = {
    "n_estimators": [200, 400],
    "max_depth": [6, 8, 10, 12],
    "min_samples_leaf": [10, 20, 50],
    "max_features": ["sqrt", 0.7]
}
```

```python
xgb_params = {
    "n_estimators": [300, 500],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.03, 0.05, 0.1],
    "subsample": [0.7, 0.8, 0.9],
    "colsample_bytree": [0.7, 0.8, 0.9]
}
```

```python
#use randomized search
from sklearn.model_selection import RandomizedSearchCV,TimeSeriesSplit
from sklearn.ensemble import RandomForestRegressor

tscv = TimeSeriesSplit(n_splits = 5)
rf =  RandomForestRegressor(random_state =42,n_jobs=-1)

rf_search = RandomizedSearchCV(
    rf,rf_params,n_iter=20,cv=tscv,scoring = "neg_mean_absolute_error",random_state =42,n_jobs=-1
)

rf_search.fit(X_train,y_train)
best_rf = rf_search.best_estimator_
```

```python
print(best_rf)
```

```python
from xgboost import XGBRegressor
from sklearn.model_selection import RandomizedSearchCV,TimeSeriesSplit

xgb = XGBRegressor(random_state = 42,tree_method = "hist",n_jobs = -1)
tscv = TimeSeriesSplit(n_splits = 5)


xgb_search = RandomizedSearchCV(
    xgb,xgb_params,n_iter=25,cv=tscv,scoring = "neg_mean_absolute_error",random_state =42,n_jobs=-1
)

xgb_search.fit(X_train,y_train)
best_xgb = xgb_search.best_estimator_
```

```python
print(best_xgb)
```

```python
#evaluated tune models
from sklearn.metrics import mean_absolute_error,r2_score

for name,model in [("RF",best_rf),("XGB",best_xgb)]:
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    print(f"{name} train MAE:", mean_absolute_error(y_train, train_pred))
    print(f"{name} test MAE :", mean_absolute_error(y_test, test_pred))
    print(f"{name} train R2 :", r2_score(y_train, train_pred))
    print(f"{name} test R2  :", r2_score(y_test, test_pred))
    print("-"*40)
    
    
```
